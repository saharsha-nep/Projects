```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Basnet\DryRun.csv")
```

```python
def GetMonthlyPoints(X):
    if X >=25000 and X <50000:
        p1 = 10
    elif X >=50000 and X <100000:
        p1 = 15
    elif X >=100000 and X <150000:
        p1 = 20
    elif X >=150000:
        p1 = 25
    else:
        p1 = 0
    return p1

def GetPaymentHistory(X):
    if X>=0.5 and X<1:
        p2 = 10
    elif X>=1 and X<2:
        p2 = 15
    elif X>=2:
        p2 = 20
    else:
        p2 = 0
    return p2

def GetOccupation(X):
    if X in ("Doctor","Engineer","Charter Accountant","Bankers","Pilot","Government Employee"):
        p3 = 20
    elif X in ("Managing Director", "business man", "Freelancer"):
        p3 = 15
    elif X == "Other":
        p3 = 10
    else:
        p3 = 0
    return p3

def GetNeaHistory(X):
    if X>=0.5 and X<1:
        p4 = 10
    elif X>=1 and X<2:
        p4 = 15
    elif X>=2:
        p4 = 20
    else:
        p4 = 0
    return p4  

def GetTopUpHistory(X):
    if X>=0.5 and X<1:
        p5 = 10
    elif X>=1 and X<2:
        p5 = 15
    elif X>=2:
        p5 = 20
    else:
        p5 = 0
    return p5  

def GetResidential(X):
    if X == True:
        p6 = 10
    else:
        p6 = 15
    return p6

def GetAge(X):
    if X>=20 and X<25:
        p7 = 10
    elif X>=25 and X<40:
        p7 = 20
    elif X>=40 and X<55:
        p7 = 30
    elif X>=55 and X<60:
        p7 = 20
    else:
        p7 = 0
    return p7
```

```python
def CreditCalculate(row):
    residential_status = row['Permanent Address'] == row['Temporary Address']
    
    P1 = GetMonthlyPoints(row['Monthly Credited Amount']) * 0.15
    P2 = GetPaymentHistory(row['Payment History Of Home']) * 0.35
    P3 = GetOccupation(row['Occupation']) * 0.15
    P4 = GetNeaHistory(row['NEA Bill Payment']) * 0.10
    P5 = GetTopUpHistory(row['TopUp Bill Payment']) * 0.05
    P6 = GetResidential(residential_status) * 0.10
    P7 = GetAge(row['Age']) * 0.10
    
    credit = P1 + P2 + P3 + P4 + P5 + P6 + P7
    return credit
```

```python
df['CreditScore'] = df.apply(CreditCalculate, axis=1)
```

```python
def EligibilityText(credit):
    if credit >= 0 and credit < 10:
        return "Not eligible, loan up to 1 Lakh"
    elif credit >= 10 and credit < 15:
        return "Medium, loan up to 2 Lakh"
    elif credit >= 15 and credit < 20:
        return "High, loan up to 5 Lakh"
    elif credit >= 20 and credit <= 22.5:
        return "Eligible, loan above 5 Lakh"
    else:
        return "Score out of range"

df['Eligibility'] = df['CreditScore'].apply(EligibilityText)
```

```python
for _, row in df.iterrows():
    print(f"{row['Name']}, {row['Eligibility']}")
```
